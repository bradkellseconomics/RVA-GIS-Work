```python
# Imports + load minimalist EDA file
import pandas as pd
from pathlib import Path

path = Path("data/analysis/servicelines_eda_minimal.csv")
df = pd.read_csv(path)

# Ensure normalized categories exist; derive if not present
def normalize_status(val):
    if pd.isna(val):
        return "Unknown"
    s = str(val).strip().lower()
    if "non" in s and "lead" in s:
        return "Non-Lead"
    if "lead" in s:
        return "Lead"
    if s in {"yes", "y"}:
        return "Lead"
    if s in {"no", "n"}:
        return "Non-Lead"
    return "Unknown"

for raw_col, norm_col in [
    ("custmaterial", "custmaterial_cat"),
    ("utilmaterial", "utilmaterial_cat"),
    ("bothsidesstatus", "bothsidesstatus_cat"),
]:
    if norm_col not in df.columns and raw_col in df.columns:
        df[norm_col] = df[raw_col].map(normalize_status)

# Summaries (counts)
order = ["Lead", "Non-Lead", "Unknown"]
counts = pd.DataFrame({
    "customer": df["custmaterial_cat"].astype(str).value_counts().reindex(order, fill_value=0),
    "utility": df["utilmaterial_cat"].astype(str).value_counts().reindex(order, fill_value=0),
    "bothsides": df["bothsidesstatus_cat"].astype(str).value_counts().reindex(order, fill_value=0),
}).astype(int)

# Optional: add totals row
counts.loc["Total"] = counts.sum(axis=0)

counts
```
